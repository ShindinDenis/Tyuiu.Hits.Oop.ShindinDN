----
<br>Создать базовый класс Event в C#, который будет представлять организованные
мероприятия. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
<b>Требования к базовому классу Event:</b>
<br><b>• Атрибуты:</b> ID мероприятия (EventId), Название мероприятия (EventName),
Дата проведения (Date).
<br><b>• Методы:</b>
<br>o RegisterParticipant(): метод для регистрации участника.
<br>o NotifyParticipants(): метод для уведомления участников о мероприятии.
<br>o GetEventDetails(): метод для получения деталей мероприятия.
<br><b>Требования к производным классам:</b>
1. Конференция (Conference): Должна содержать дополнительные атрибуты,
такие как Количество спикеров (SpeakersCount).
Метод NotifyParticipants() должен быть переопределен для отправки
уведомлений с программой конференции.
2. Фестиваль (Festival): Должна содержать дополнительные атрибуты, такие как
Количество музыкальных групп (BandsCount).
Метод RegisterParticipant() должен быть переопределен для регистрации
участников с выбором интересующих музыкальных жанров.
3. Семинар (Seminar) (если требуется третий класс): Должна содержать
дополнительные атрибуты, такие как Тематические области (Topics).
Метод GetEventDetails() должен быть переопределен для отображения
тематических областей семинара вместе с другими деталями мероприятия.

```
using System;
using System.Collections.Generic;
using System.Linq;


public delegate bool EventFilterDelegate(Event e);

public delegate void EventStatusHandler(Event sender, string message);

public interface INotificationService
{
    void SendNotification(string recipient, string message);
}

public interface ITickeable
{
    decimal GetPrice();
    void SellTicket(string participant);
    int GetTicketsSold();
}

public class EmailNotificationService : INotificationService
{
    public void SendNotification(string recipient, string message)
    {
        Console.WriteLine($"[EMAIL] Отправлено {recipient}: \"{message}\"");
    }
}

public class SmsNotificationService : INotificationService
{
    public void SendNotification(string recipient, string message)
    {
        Console.WriteLine($"[SMS] Отправлено {recipient}: \"{message}\"");
    }
}

public class Event : ITickeable
{
    public int EventId { get; set; }
    public string EventName { get; set; }
    public DateTime Date { get; set; }
    public string Location { get; set; }
    protected List<string> participants = new List<string>();
    public decimal TicketPrice { get; set; }
    public string Organizer { get; set; }
    public string EventType { get; protected set; } 
    private readonly INotificationService _notificationService;
    private bool IsCancelled { get; set; }

    public event EventStatusHandler ParticipantRegistered;
    public event EventStatusHandler EventCancelled;

    public Event(int eventId, string eventName, DateTime date, string location, decimal ticketPrice, string organizer, INotificationService notificationService)
    {
        EventId = eventId;
        EventName = eventName;
        Date = date;
        Location = location;
        TicketPrice = ticketPrice;
        Organizer = organizer;
        EventType = "General";
        IsCancelled = false;
        _notificationService = notificationService;
    }

    public void AnnounceTo(Event otherEvent)
    {
        Console.WriteLine($"\nМероприятие '{EventName}' объявляет о сотрудничестве с '{otherEvent.EventName}'.");
    }
    
    public virtual void RegisterParticipant(string participant)
    {
        if (IsCancelled)
        {
            Console.WriteLine($"Ошибка: Мероприятие '{EventName}' отменено, регистрация невозможна.");
            return;
        }
        participants.Add(participant);
        Console.WriteLine($"Участник {participant} успешно зарегистрирован на мероприятие '{EventName}'.");
        ParticipantRegistered?.Invoke(this, $"Новый участник: {participant}.");
    }

    public virtual void CancelEvent()
    {
        if (!IsCancelled)
        {
            IsCancelled = true;
            Console.WriteLine($"\n!!! ВАЖНО: Мероприятие '{EventName}' ОТМЕНЕНО.");
            NotifyParticipants($"СРОЧНО: Мероприятие '{EventName}' отменено.");
            EventCancelled?.Invoke(this, "Мероприятие было отменено!");
        }
    }
    
    public void SetNewDate(DateTime newDate)
    {
        Date = newDate;
        Console.WriteLine($"\n[INFO] Дата мероприятия '{EventName}' изменена на {newDate.ToShortDateString()}.");
    }
    
    public string GetOrganizerInfo()
    {
        return $"Мероприятие '{EventName}' организовано: {Organizer}.";
    }

    public void NotifyParticipants(string message)
    {
        Console.WriteLine($"\nОтправка кастомного уведомления для {participants.Count} участников '{EventName}':");
        foreach (var p in participants)
        {
            _notificationService.SendNotification(p, message);
        }
    }
    
    public virtual void GetEventDetails()
    {
        Console.WriteLine($"\n--- Детали мероприятия ({EventType}) ---");
        Console.WriteLine($"ID: {EventId}");
        Console.WriteLine($"Название: {EventName}");
        Console.WriteLine($"Дата: {Date.ToShortDateString()}");
        Console.WriteLine($"Место: {Location}");
        Console.WriteLine($"Организатор: {Organizer}");
        Console.WriteLine($"Цена билета: {TicketPrice:N2} ₽");
        Console.WriteLine($"Статус: {(IsCancelled ? "ОТМЕНЕНО" : "Активно")}");
        Console.WriteLine($"Количество зарегистрированных участников: {participants.Count}");
    }

    decimal ITickeable.GetPrice() => this.TicketPrice;
    int ITickeable.GetTicketsSold() => this.participants.Count;
    void ITickeable.SellTicket(string participant)
    {
        if (TicketPrice > 0)
        {
            Console.WriteLine($"[TicketSystem] Продажа билета для {participant} на '{EventName}' за {TicketPrice:N2} ₽...");
            Console.WriteLine($"[TicketSystem] Оплата получена.");
        }
        else
        {
            Console.WriteLine($"[TicketSystem] Регистрация (бесплатно) для {participant} на '{EventName}'.");
        }
        this.RegisterParticipant(participant);
    }
}

public class Conference : Event
{
    public int SpeakersCount { get; set; }
    public string MainTopic { get; set; }
    public List<string> Sponsors { get; private set; }
    public Dictionary<TimeSpan, string> Schedule { get; private set; }
    public bool HasNetworkingEvent { get; set; }

    public Conference(int eventId, string eventName, DateTime date, string location, int speakersCount, string mainTopic, decimal ticketPrice, string organizer, INotificationService notificationService)
        : base(eventId, eventName, date, location, ticketPrice, organizer, notificationService)
    {
        SpeakersCount = speakersCount;
        MainTopic = mainTopic;
        EventType = "Conference"; 
        Sponsors = new List<string>();
        Schedule = new Dictionary<TimeSpan, string>();
    }
    
    public void AddSponsor(string sponsorName)
    {
        Sponsors.Add(sponsorName);
        Console.WriteLine($"[Sponsor] {sponsorName} теперь спонсор конференции '{EventName}'.");
    }

    public void AddScheduleItem(TimeSpan time, string description)
    {
        Schedule[time] = description;
        Console.WriteLine($"[Schedule] Добавлено в расписание {time}: {description}");
    }
    
    public void PrintSchedule()
    {
        Console.WriteLine($"\n--- Расписание для '{EventName}' ---");
        foreach (var item in Schedule.OrderBy(kv => kv.Key))
        {
            Console.WriteLine($"{item.Key} - {item.Value}");
        }
        Console.WriteLine("---------------------------------");
    }
    
    public void ExchangeSpeakers(Conference otherConference)
    {
        Console.WriteLine($"\nКонференция '{EventName}' обменивается {SpeakersCount} спикерами с конференцией '{otherConference.EventName}' у которой {otherConference.SpeakersCount} спикеров.");
    }
    
    public override void CancelEvent()
    {
        base.CancelEvent();
        Console.WriteLine($"Спикерам и спонсорам отправлено уведомление об отмене конференции.");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Количество спикеров: {SpeakersCount}");
        Console.WriteLine($"Основная тема: {MainTopic}");
        Console.WriteLine($"Нетворкинг: {(HasNetworkingEvent ? "Запланирован" : "Нет")}");
        if (Sponsors.Count > 0)
        {
            Console.WriteLine($"Спонсоры: {string.Join(", ", Sponsors)}");
        }
    }
}

public class Festival : Event
{
    public int BandsCount { get; set; }
    public bool HasFireworks { get; set; }
    public string MusicGenre { get; set; }
    public int AgeRating { get; set; }
    public int FoodTrucksCount { get; private set; }

    public Festival(int eventId, string eventName, DateTime date, string location, int bandsCount, bool hasFireworks, decimal ticketPrice, string organizer, string genre, int ageRating, INotificationService notificationService)
        : base(eventId, eventName, date, location, ticketPrice, organizer, notificationService)
    {
        BandsCount = bandsCount;
        HasFireworks = hasFireworks;
        MusicGenre = genre;
        AgeRating = ageRating;
        FoodTrucksCount = 0;
        EventType = "Festival";
    }
    
    public void AddFoodTruck()
    {
        FoodTrucksCount++;
        Console.WriteLine($"[Info] На фестивале '{EventName}' теперь {FoodTrucksCount} фуд-траков.");
    }
    
    public bool CheckAge(int participantAge)
    {
        bool allowed = participantAge >= AgeRating;
        Console.WriteLine($"[Age Check] Возраст {participantAge}. Ограничение {AgeRating}+. Доступ: {(allowed ? "Разрешен" : "Запрещен")}");
        return allowed;
    }
    
    public void PerformWith(Festival otherFestival)
    {
        Console.WriteLine($"\nФестиваль '{EventName}' и фестиваль '{otherFestival.EventName}' объединяют свои {BandsCount} и {otherFestival.BandsCount} группы для совместного выступления.");
    }

    public void RegisterParticipant(string participant, string favoriteGenre)
    {
        base.RegisterParticipant(participant); 
        Console.WriteLine($"Любимый музыкальный жанр: {favoriteGenre}.");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Музыкальный жанр: {MusicGenre}");
        Console.WriteLine($"Количество музыкальных групп: {BandsCount}");
        Console.WriteLine($"Будет фейерверк: {(HasFireworks ? "Да" : "Нет")}");
        Console.WriteLine($"Возрастной рейтинг: {AgeRating}+");
        Console.WriteLine($"Кол-во фуд-траков: {FoodTrucksCount}");
    }
}

public class Seminar : Event
{
    public List<string> Topics { get; set; }
    public int MaxParticipants { get; set; }
    public string SpeakerName { get; set; }
    public double DurationHours { get; set; }
    public List<string> RequiredMaterials { get; set; }

    public Seminar(int eventId, string eventName, DateTime date, string location, List<string> topics, int maxParticipants, decimal ticketPrice, string organizer, string speaker, double duration, INotificationService notificationService)
        : base(eventId, eventName, date, location, ticketPrice, organizer, notificationService)
    {
        Topics = topics;
        MaxParticipants = maxParticipants;
        SpeakerName = speaker;
        DurationHours = duration;
        RequiredMaterials = new List<string>();
        EventType = "Seminar";
    }
    
    public void AddRequiredMaterial(string material)
    {
        RequiredMaterials.Add(material);
        Console.WriteLine($"[Info] Для семинара '{EventName}' теперь требуется: {material}.");
    }

    public void GetRequiredMaterialsList()
    {
        Console.WriteLine($"\nМатериалы для семинара '{EventName}':");
        if (!RequiredMaterials.Any()) Console.WriteLine("- Ничего не требуется.");
        else RequiredMaterials.ForEach(m => Console.WriteLine($"- {m}")); // ИСПРАВЛЕНО ЗДЕСЬ
    }
    
    public string GetDuration() => $"Семинар '{EventName}' продлится {DurationHours} ч.";
    
    public void CollaborateOnTopic(Seminar otherSeminar)
    {
        Console.WriteLine($"\nСеминар '{EventName}' и семинар '{otherSeminar.EventName}' начинают совместную работу над темой '{Topics.FirstOrDefault()}'.");
    }

    public override void RegisterParticipant(string participant)
    {
        if (participants.Count >= MaxParticipants)
        {
             Console.WriteLine($"Ошибка: Семинар '{EventName}' полон ({MaxParticipants} участников).");
             return;
        }
        base.RegisterParticipant(participant);
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Спикер: {SpeakerName}");
        Console.WriteLine($"Продолжительность: {DurationHours} ч.");
        Console.WriteLine($"Макс. участников: {MaxParticipants}");
        Console.WriteLine("Темы семинара:");
        Topics.ForEach(t => Console.WriteLine($"- {t}"));
        GetRequiredMaterialsList();
    }
}

public class EventManager
{
    private List<Event> _events = new List<Event>();
    public string ManagerName { get; private set; }

    public EventManager(string name)
    {
        ManagerName = name;
    }

    public void AddEvent(Event eventItem)
    {
        _events.Add(eventItem);
    }
    
    public List<Event> FindEvents(EventFilterDelegate filter)
    {
        Console.WriteLine($"\n[EventManager] Поиск мероприятий по фильтру...");
        return _events.Where(e => filter(e)).ToList();
    }

    public void RunAllEventDetails()
    {
        Console.WriteLine($"\n\n*** Менеджер {ManagerName}: Запуск отчетов (Полиморфизм) ***");
        foreach (var eventItem in _events)
        {
            eventItem.GetEventDetails(); 
        }
    }
}



INotificationService emailService = new EmailNotificationService();
INotificationService smsService = new SmsNotificationService();

void LogParticipant(Event sender, string message)
{
    Console.WriteLine($"[EVENT_LOG] От '{sender.EventName}': {message}");
}
void AlertManager(Event sender, string message)
{
    Console.WriteLine($"[ALERT!!!] От '{sender.EventName}': {message} Требуется вмешательство!");
}

var generalEvent = new Event(101, "Общее собрание", new DateTime(2025, 11, 25), "Офис A", 0, "HR Отдел", emailService);
var techConference = new Conference(201, "Технологический форум", new DateTime(2025, 12, 10), "Павильон №1", 15, "ИИ и Автоматизация", 1500, "TechOrg", emailService);
var musicFestival = new Festival(301, "Summer Jam", new DateTime(2026, 6, 15), "Парк Горького", 20, true, 5000, "MusicFest Ltd.", "Rock", 18, smsService);
var codingSeminar = new Seminar(401, "Мастер-класс по C#", new DateTime(2026, 1, 20), "Онлайн", new List<string> { "ООП", "LINQ", "Асинхронное программирование" }, 5, 1000, "IT-Univer", "Алексей Иванов", 4.5, emailService);
var securityConference = new Conference(202, "Конференция по кибербезопасности", new DateTime(2025, 12, 12), "Павильон №2", 10, "Защита данных", 2000, "SecureInc", emailService);
var rockFestival = new Festival(302, "Rock Fest", new DateTime(2026, 7, 20), "Стадион", 12, false, 4500, "MusicFest Ltd.", "Alternative", 16, smsService);
var pythonSeminar = new Seminar(402, "Семинар по Python", new DateTime(2026, 2, 15), "Класс B", new List<string> { "Django", "Flask" }, 15, 0, "IT-Univer", "Мария Петрова", 3, emailService);


Console.WriteLine("\n--- Подписка на события ---");
techConference.ParticipantRegistered += LogParticipant;
techConference.EventCancelled += AlertManager;
musicFestival.ParticipantRegistered += LogParticipant;


Console.WriteLine("\n--- Начальная настройка мероприятий ---");
techConference.AddSponsor("МегаКорп");
techConference.AddScheduleItem(new TimeSpan(9, 0, 0), "Регистрация и кофе");
techConference.AddScheduleItem(new TimeSpan(10, 0, 0), "Keynote: Будущее ИИ");
techConference.PrintSchedule();

musicFestival.AddFoodTruck();
musicFestival.AddFoodTruck();
musicFestival.CheckAge(20); 
musicFestival.CheckAge(16); 

codingSeminar.AddRequiredMaterial("Visual Studio 2022");
codingSeminar.AddRequiredMaterial("Аккаунт .NET");
Console.WriteLine(codingSeminar.GetDuration());

Console.WriteLine(generalEvent.GetOrganizerInfo());
generalEvent.SetNewDate(new DateTime(2025, 11, 26)); 
generalEvent.RegisterParticipant("Иван");


Console.WriteLine("\n----------------------------------");
ITickeable ticketedFestival = musicFestival;
Console.WriteLine($"Цена билета на {musicFestival.EventName}: {ticketedFestival.GetPrice():N2} ₽");
ticketedFestival.SellTicket("Александр"); 
ticketedFestival.SellTicket("Елена");
Console.WriteLine($"Билетов продано (через интерфейс): {ticketedFestival.GetTicketsSold()}");

ITickeable freeSeminar = pythonSeminar;
Console.WriteLine($"\nЦена билета на {pythonSeminar.EventName}: {freeSeminar.GetPrice():N2} ₽");
freeSeminar.SellTicket("Виктор");


Console.WriteLine("\n----------------------------------");
techConference.RegisterParticipant("Анна");
techConference.NotifyParticipants($"Подготовлена программа конференции '{techConference.EventName}' с {techConference.SpeakersCount} спикерами по теме '{techConference.MainTopic}'.");

musicFestival.RegisterParticipant("Сергей", "Рок");
musicFestival.NotifyParticipants($"Напоминаем о мероприятии '{musicFestival.EventName}' {musicFestival.Date.ToShortDateString()}");

codingSeminar.RegisterParticipant("Мария");


Console.WriteLine("\n----------------------------------");
generalEvent.AnnounceTo(techConference);
techConference.ExchangeSpeakers(securityConference);
musicFestival.PerformWith(rockFestival);
codingSeminar.CollaborateOnTopic(pythonSeminar);


var manager = new EventManager("Глобальный");
manager.AddEvent(techConference);
manager.AddEvent(musicFestival);
manager.AddEvent(codingSeminar);
manager.AddEvent(generalEvent);
manager.AddEvent(securityConference);
manager.AddEvent(rockFestival);
manager.AddEvent(pythonSeminar);

Console.WriteLine("\n----------------------------------");
var expensiveEvents = manager.FindEvents(e => e.TicketPrice >= 2000 && e is Festival);
Console.WriteLine("Найдены дорогие фестивали:");
expensiveEvents.ForEach(e => Console.WriteLine($"- {e.EventName}"));

Console.WriteLine("\n----------------------------------");
var freeOrOnlineEvents = manager.FindEvents(e => e.TicketPrice == 0 || e.Location == "Онлайн");
Console.WriteLine("Найдены бесплатные или онлайн-мероприятия:");
freeOrOnlineEvents.ForEach(e => Console.WriteLine($"- {e.EventName} (Цена: {e.TicketPrice:N2}, Место: {e.Location})"));


Console.WriteLine("\n----------------------------------");
techConference.CancelEvent(); 

manager.RunAllEventDetails();
```
