----
<br>Создать базовый класс Event в C#, который будет представлять организованные
мероприятия. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
<b>Требования к базовому классу Event:</b>
<br><b>• Атрибуты:</b> ID мероприятия (EventId), Название мероприятия (EventName),
Дата проведения (Date).
<br><b>• Методы:</b>
<br>o RegisterParticipant(): метод для регистрации участника.
<br>o NotifyParticipants(): метод для уведомления участников о мероприятии.
<br>o GetEventDetails(): метод для получения деталей мероприятия.
<br><b>Требования к производным классам:</b>
1. Конференция (Conference): Должна содержать дополнительные атрибуты,
такие как Количество спикеров (SpeakersCount).
Метод NotifyParticipants() должен быть переопределен для отправки
уведомлений с программой конференции.
2. Фестиваль (Festival): Должна содержать дополнительные атрибуты, такие как
Количество музыкальных групп (BandsCount).
Метод RegisterParticipant() должен быть переопределен для регистрации
участников с выбором интересующих музыкальных жанров.
3. Семинар (Seminar) (если требуется третий класс): Должна содержать
дополнительные атрибуты, такие как Тематические области (Topics).
Метод GetEventDetails() должен быть переопределен для отображения
тематических областей семинара вместе с другими деталями мероприятия.

```
public interface IOnlineEvent
{
    void StartOnlineStream();
    void StopOnlineStream();
}

public interface ISponsorable
{
    void AddSponsor(string sponsor);
    void ShowSponsors();
}
public class Event
{
    public int EventId { get; set; }
    public string EventName { get; set; }
    public DateTime Date { get; set; }
    public string Location { get; set; }
    public decimal TicketPrice { get; set; }
    public bool IsCancelled { get; private set; }

    protected List<string> participants = new List<string>();

    public Event(int eventId, string eventName, DateTime date, string location = "Не указан", decimal ticketPrice = 0)
    {
        EventId = eventId;
        EventName = eventName;
        Date = date;
        Location = location;
        TicketPrice = ticketPrice;
        IsCancelled = false;
    }

    public virtual void RegisterParticipant(string participant)
    {
        participants.Add(participant);
        Console.WriteLine($"Участник {participant} зарегистрирован на '{EventName}'.");
    }

    public virtual void NotifyParticipants()
    {
        Console.WriteLine($"Уведомление для {participants.Count} участников мероприятия '{EventName}'.");
    }

    public virtual void GetEventDetails()
    {
        Console.WriteLine($"\n--- Детали мероприятия ---");
        Console.WriteLine($"ID: {EventId}");
        Console.WriteLine($"Название: {EventName}");
        Console.WriteLine($"Дата: {Date.ToShortDateString()}");
        Console.WriteLine($"Место проведения: {Location}");
        Console.WriteLine($"Стоимость билета: {TicketPrice} руб.");
        Console.WriteLine($"Количество зарегистрированных участников: {participants.Count}");
    }

    public void CancelEvent()
    {
        IsCancelled = true;
        Console.WriteLine($"Мероприятие '{EventName}' было отменено.");
    }

    public void AnnounceTo(Event otherEvent)
    {
        Console.WriteLine($"\nМероприятие '{EventName}' объявляет о сотрудничестве с '{otherEvent.EventName}'.");
    }
}
public class Conference : Event, ISponsorable
{
    public int SpeakersCount { get; set; }
    public string Organizer { get; set; }
    public string Topic { get; set; }

    private List<string> sponsors = new List<string>();

    public Conference(int eventId, string eventName, DateTime date, int speakersCount,
                      string organizer = "Неизвестный", string topic = "Общая тема",
                      string location = "Не указан", decimal ticketPrice = 0)
        : base(eventId, eventName, date, location, ticketPrice)
    {
        SpeakersCount = speakersCount;
        Organizer = organizer;
        Topic = topic;
    }

    public override void NotifyParticipants()
    {
        Console.WriteLine($"\nУведомление для участников конференции '{EventName}':");
        Console.WriteLine($"Тема: {Topic}. Подготовлено {SpeakersCount} спикеров.");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Организатор: {Organizer}");
        Console.WriteLine($"Количество спикеров: {SpeakersCount}");
        Console.WriteLine($"Тема конференции: {Topic}");
    }

    public void ExchangeSpeakers(Conference otherConference)
    {
        Console.WriteLine($"\nКонференция '{EventName}' обменивается {SpeakersCount} спикерами с '{otherConference.EventName}'.");
    }

    public void AddSponsor(string sponsor)
    {
        sponsors.Add(sponsor);
        Console.WriteLine($"Спонсор {sponsor} добавлен для конференции '{EventName}'.");
    }

    public void ShowSponsors()
    {
        Console.WriteLine($"Спонсоры конференции '{EventName}': {string.Join(", ", sponsors)}");
    }
}

public class InternationalConference : Conference, IOnlineEvent
{
    public string Country { get; set; }
    public string Language { get; set; }

    public InternationalConference(int eventId, string eventName, DateTime date, int speakersCount,
                                   string organizer, string topic, string country, string language,
                                   string location = "Не указан", decimal ticketPrice = 0)
        : base(eventId, eventName, date, speakersCount, organizer, topic, location, ticketPrice)
    {
        Country = country;
        Language = language;
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Страна проведения: {Country}");
        Console.WriteLine($"Основной язык: {Language}");
    }

    public void StartOnlineStream()
    {
        Console.WriteLine($"Онлайн-трансляция международной конференции '{EventName}' началась.");
    }

    public void StopOnlineStream()
    {
        Console.WriteLine($"Онлайн-трансляция международной конференции '{EventName}' завершена.");
    }
}

public class Festival : Event
{
    public int BandsCount { get; set; }
    public string MainGenre { get; set; }
    public string Sponsor { get; set; }

    public Festival(int eventId, string eventName, DateTime date, int bandsCount,
                    string mainGenre = "Разное", string sponsor = "Нет",
                    string location = "Не указан", decimal ticketPrice = 0)
        : base(eventId, eventName, date, location, ticketPrice)
    {
        BandsCount = bandsCount;
        MainGenre = mainGenre;
        Sponsor = sponsor;
    }

    public void RegisterParticipant(string participant, string favoriteGenre)
    {
        participants.Add(participant);
        Console.WriteLine($"\nУчастник {participant} зарегистрирован на фестиваль '{EventName}'. Любимый жанр: {favoriteGenre}");
    }

    public override void RegisterParticipant(string participant)
    {
        RegisterParticipant(participant, "Не указан");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Количество групп: {BandsCount}");
        Console.WriteLine($"Основной жанр: {MainGenre}");
        Console.WriteLine($"Спонсор: {Sponsor}");
    }

    public void PerformWith(Festival otherFestival)
    {
        Console.WriteLine($"\nФестиваль '{EventName}' и '{otherFestival.EventName}' объединяют {BandsCount} и {otherFestival.BandsCount} групп.");
    }
}

public class Seminar : Event
{
    public List<string> Topics { get; set; }
    public string Lecturer { get; set; }
    public int DurationHours { get; set; }

    public Seminar(int eventId, string eventName, DateTime date, List<string> topics,
                   string lecturer = "Приглашенный эксперт", int durationHours = 2,
                   string location = "Не указан", decimal ticketPrice = 0)
        : base(eventId, eventName, date, location, ticketPrice)
    {
        Topics = topics;
        Lecturer = lecturer;
        DurationHours = durationHours;
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Лектор: {Lecturer}");
        Console.WriteLine($"Продолжительность: {DurationHours} часов");
        Console.WriteLine("Темы семинара:");
        foreach (var topic in Topics)
        {
            Console.WriteLine($"- {topic}");
        }
    }

    public void CollaborateOnTopic(Seminar otherSeminar)
    {
        Console.WriteLine($"\nСеминар '{EventName}' и '{otherSeminar.EventName}' начинают совместную работу над темой '{Topics[0]}'.");
    }
}

        Event generalEvent = new Event(101, "Общее собрание", new DateTime(2023, 11, 25), "Москва", 500);
        Conference techConference = new Conference(201, "Технологический форум", new DateTime(2023, 12, 10), 15,
                                                   "TechOrg", "AI и будущее", "Санкт-Петербург", 1500);
        Festival musicFestival = new Festival(301, "Summer Jam", new DateTime(2024, 6, 15), 20,
                                              "Pop", "Pepsi", "Сочи", 1200);
        Seminar codingSeminar = new Seminar(401, "Мастер-класс по C#", new DateTime(2024, 1, 20),
                                            new List<string> { "ООП", "LINQ", "Async" },
                                            "Иван Петров", 4, "Новосибирск", 800);

        Conference securityConference = new Conference(202, "Конференция по кибербезопасности", new DateTime(2023, 12, 12), 10,
                                                       "SecOrg", "Киберугрозы", "Москва", 1300);
        Festival rockFestival = new Festival(302, "Rock Fest", new DateTime(2024, 7, 20), 12,
                                             "Rock", "Coca-Cola", "Екатеринбург", 1000);
        Seminar pythonSeminar = new Seminar(402, "Семинар по Python", new DateTime(2024, 2, 15),
                                            new List<string> { "Django", "Flask" },
                                            "Алексей Иванов", 3, "Казань", 700);

        InternationalConference intlConf = new InternationalConference(501, "Международный форум", new DateTime(2024, 5, 10),
                                                                       25, "GlobalOrg", "Мировые технологии",
                                                                       "Германия", "Английский", "Берлин", 5000);

        generalEvent.RegisterParticipant("Иван");
        generalEvent.GetEventDetails();
        generalEvent.NotifyParticipants();

        Console.WriteLine("\n----------------------------------\n");

        techConference.RegisterParticipant("Анна");
        techConference.AddSponsor("Яндекс");
        techConference.ShowSponsors();
        techConference.GetEventDetails();
        techConference.NotifyParticipants();

        Console.WriteLine("\n----------------------------------\n");

        musicFestival.RegisterParticipant("Сергей", "Рок");
        musicFestival.GetEventDetails();

        Console.WriteLine("\n----------------------------------\n");

        codingSeminar.RegisterParticipant("Мария");
        codingSeminar.GetEventDetails();

        Console.WriteLine("\n----------------------------------\n");

        intlConf.RegisterParticipant("Дмитрий");
        intlConf.AddSponsor("Сбер");
        intlConf.GetEventDetails();
        intlConf.StartOnlineStream();
        intlConf.StopOnlineStream();

        generalEvent.AnnounceTo(techConference);
        techConference.ExchangeSpeakers(securityConference);
        musicFestival.PerformWith(rockFestival);
        codingSeminar.CollaborateOnTopic(pythonSeminar);
```
