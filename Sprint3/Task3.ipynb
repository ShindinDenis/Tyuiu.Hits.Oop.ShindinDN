<b>Требования к базовому классу Event:</b>

<br><b>• Методы:</b>
<br>o RegisterParticipant(): метод для регистрации участника.
<br>o NotifyParticipants(): метод для уведомления участников о мероприятии.
<br>o GetEventDetails(): метод для получения деталей мероприятия.

Количество музыкальных групп (BandsCount).

участников с выбором интересующих музыкальных жанров.
3. Семинар (Seminar) (если требуется третий класс): Должна содержать
дополнительные атрибуты, такие как Тематические области (Topics).
Метод GetEventDetails() должен быть переопределен для отображения
тематических областей семинара вместе с другими деталями мероприятия.

```
public class Event
{
    public int EventId { get; set; }
    public string EventName { get; set; }
    public DateTime Date { get; set; }
    
    public string Location { get; set; }
    private bool IsCancelled { get; set; }
    
    protected List<string> participants = new List<string>();

    public Event(int eventId, string eventName, DateTime date, string location)
    {
        EventId = eventId;
        EventName = eventName;
        Date = date;
        Location = location; 
        IsCancelled = false;
    }

    public virtual void RegisterParticipant(string participant)
    {
        if (IsCancelled)
        {
            Console.WriteLine($"Ошибка: Мероприятие '{EventName}' отменено, регистрация невозможна.");
            return;
        }
        participants.Add(participant);
        Console.WriteLine($"Участник {participant} успешно зарегистрирован на мероприятие '{EventName}'.");
    }

    public virtual void NotifyParticipants()
    {
        Console.WriteLine($"Уведомление для всех {participants.Count} участников мероприятия '{EventName}'.");
    }

    public virtual void GetEventDetails()
    {
        Console.WriteLine($"\n--- Детали мероприятия ---");
        Console.WriteLine($"ID: {EventId}");
        Console.WriteLine($"Название: {EventName}");
        Console.WriteLine($"Дата: {Date.ToShortDateString()}");
        Console.WriteLine($"Место: {Location}");
        Console.WriteLine($"Статус: {(IsCancelled ? "ОТМЕНЕНО" : "Активно")}");
        Console.WriteLine($"Количество зарегистрированных участников: {participants.Count}");
    }
    public virtual void CancelEvent()
    {
        if (!IsCancelled)
        {
            IsCancelled = true;
            Console.WriteLine($"\n!!! ВАЖНО: Мероприятие '{EventName}' ОТМЕНЕНО.");
        }
    }


    public void CheckCapacity(int maxCapacity)
    {
        CheckCapacity(participants.Count, maxCapacity);
    }
    
    public void CheckCapacity(int current, int max)
    {
        if (current >= max)
        {
            Console.WriteLine($"[Capacity Check] Внимание! Мероприятие '{EventName}' достигло максимальной вместимости ({max}).");
        }
        else
        {
            Console.WriteLine($"[Capacity Check] Осталось мест: {max - current}.");
        }
    }

    public void AnnounceTo(Event otherEvent)
    {
        Console.WriteLine($"\nМероприятие '{EventName}' объявляет о сотрудничестве с '{otherEvent.EventName}'.");
    }
}


public class Conference : Event
{
    public int SpeakersCount { get; set; }
    public string MainTopic { get; set; }

    public Conference(int eventId, string eventName, DateTime date, string location, int speakersCount, string mainTopic)
        : base(eventId, eventName, date, location)
    {
        SpeakersCount = speakersCount;
        MainTopic = mainTopic;
    }

    public override void NotifyParticipants()
    {
        Console.WriteLine($"\nУведомление для участников конференции '{EventName}':");
        Console.WriteLine($"Подготовлена программа с {SpeakersCount} спикерами по теме '{MainTopic}'.");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Количество спикеров: {SpeakersCount}");
        Console.WriteLine($"Основная тема: {MainTopic}");
    }
    
    public override void CancelEvent()
    {
        base.CancelEvent();
        Console.WriteLine($"Спикерам отправлено уведомление об отмене конференции.");
    }
    
    public void ExchangeSpeakers(Conference otherConference)
    {
        Console.WriteLine($"\nКонференция '{EventName}' обменивается {SpeakersCount} спикерами с конференцией '{otherConference.EventName}' у которой {otherConference.SpeakersCount} спикеров.");
    }
}


public class Festival : Event
{
    public int BandsCount { get; set; }
    public bool HasFireworks { get; set; }

    public Festival(int eventId, string eventName, DateTime date, string location, int bandsCount, bool hasFireworks)
        : base(eventId, eventName, date, location)
    {
        BandsCount = bandsCount;
        HasFireworks = hasFireworks;
    }

    public void RegisterParticipant(string participant, string favoriteGenre)
    {
        base.RegisterParticipant(participant); 
        Console.WriteLine($"Любимый музыкальный жанр: {favoriteGenre}.");
    }

    public override void RegisterParticipant(string participant)
    {
        RegisterParticipant(participant, "Не указан");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Количество музыкальных групп: {BandsCount}");
        Console.WriteLine($"Будет фейерверк: {(HasFireworks ? "Да" : "Нет")}");
    }
    
    public void PerformWith(Festival otherFestival)
    {
        Console.WriteLine($"\nФестиваль '{EventName}' и фестиваль '{otherFestival.EventName}' объединяют свои {BandsCount} и {otherFestival.BandsCount} группы для совместного выступления.");
    }
}


public class Seminar : Event
{
    public List<string> Topics { get; set; }
    public int MaxParticipants { get; set; }

    public Seminar(int eventId, string eventName, DateTime date, string location, List<string> topics, int maxParticipants)
        : base(eventId, eventName, date, location)
    {
        Topics = topics;
        MaxParticipants = maxParticipants;
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Макс. участников: {MaxParticipants}");
        Console.WriteLine("Темы семинара:");
        foreach (var topic in Topics)
        {
            Console.WriteLine($"- {topic}");
        }
    }
    
    public override void RegisterParticipant(string participant)
    {
        if (participants.Count >= MaxParticipants)
        {
             Console.WriteLine($"Ошибка: Семинар '{EventName}' полон ({MaxParticipants} участников).");
             return;
        }
        base.RegisterParticipant(participant);
    }
    
    public void CollaborateOnTopic(Seminar otherSeminar)
    {
        Console.WriteLine($"\nСеминар '{EventName}' и семинар '{otherSeminar.EventName}' начинают совместную работу над темой '{Topics[0]}'.");
    }
}

public class EventManager<T> where T : Event
{
    private List<T> _events = new List<T>();
    public string ManagerName { get; private set; }

    public EventManager(string name)
    {
        ManagerName = name;
    }

    public void AddEvent(T eventItem)
    {
        _events.Add(eventItem);
    }

    public void RunAllEventDetails()
    {
        Console.WriteLine($"\n\n*** Менеджер {ManagerName}: Запуск отчетов (Полиморфизм) ***");
        foreach (var eventItem in _events)
        {
            eventItem.GetEventDetails(); 
        }
    }
}


        Event generalEvent = new Event(101, "Общее собрание", new DateTime(2023, 11, 25), "Офис A");
        Conference techConference = new Conference(201, "Технологический форум", new DateTime(2023, 12, 10), "Павильон №1", 15, "ИИ и Автоматизация");
        Festival musicFestival = new Festival(301, "Summer Jam", new DateTime(2024, 6, 15), "Парк Горького", 20, true);
        Seminar codingSeminar = new Seminar(401, "Мастер-класс по C#", new DateTime(2024, 1, 20), "Онлайн", new List<string> { "ООП", "LINQ", "Асинхронное программирование" }, 5);
        
        Conference securityConference = new Conference(202, "Конференция по кибербезопасности", new DateTime(2023, 12, 12), "Павильон №2", 10, "Защита данных");
        Festival rockFestival = new Festival(302, "Rock Fest", new DateTime(2024, 7, 20), "Стадион", 12, false);
        Seminar pythonSeminar = new Seminar(402, "Семинар по Python", new DateTime(2024, 2, 15), "Класс B", new List<string> { "Django", "Flask" }, 15);

        
        generalEvent.RegisterParticipant("Иван");
        generalEvent.GetEventDetails();
        generalEvent.NotifyParticipants();
        
        Console.WriteLine("\n----------------------------------");
        
        techConference.RegisterParticipant("Анна");
        techConference.GetEventDetails();
        techConference.NotifyParticipants();

        Console.WriteLine("\n----------------------------------");

        musicFestival.RegisterParticipant("Сергей", "Рок"); 
        musicFestival.GetEventDetails();
        musicFestival.NotifyParticipants();

        Console.WriteLine("\n----------------------------------");

        codingSeminar.RegisterParticipant("Мария");
        codingSeminar.GetEventDetails();
        codingSeminar.NotifyParticipants();
        
        generalEvent.AnnounceTo(techConference);
        techConference.ExchangeSpeakers(securityConference);
        musicFestival.PerformWith(rockFestival);
        codingSeminar.CollaborateOnTopic(pythonSeminar);

        
        techConference.CancelEvent(); 
        generalEvent.CancelEvent(); 
        
        generalEvent.RegisterParticipant("Петр");
        
        Console.WriteLine("\n--- Демонстрация перегрузки CheckCapacity ---");
        techConference.CheckCapacity(20); 
        codingSeminar.CheckCapacity(10, 15);
        
        EventManager<Event> manager = new EventManager<Event>("Глобальный");
        manager.AddEvent(techConference);
        manager.AddEvent(musicFestival);
        manager.AddEvent(codingSeminar);
        
        manager.RunAllEventDetails();
```
